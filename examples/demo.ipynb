## Images in the terminal

This PNG is created with the standard library. The preview uses terminal colors; **o** opens the original resolution in your browser.

```python
import math, struct, zlib
from IPython.display import display, Image, HTML

w, h = 320, 100
def chunk(kind, data):
    return struct.pack(">I", len(data)) + kind + data + struct.pack(">I", zlib.crc32(kind + data) & 0xffffffff)
rows = []
for y in range(h):
    row = bytearray([0])
    for x in range(w):
        curve = int(h / 2 + math.sin(x / 20) * 35)
        row.extend((86, 211, 183) if abs(y - curve) < 2 else (22, 29, 42))
    rows.append(row)
png = b"\x89PNG\r\n\x1a\n" + chunk(b"IHDR", struct.pack(">IIBBBBB", w, h, 8, 2, 0, 0, 0))
png += chunk(b"IDAT", zlib.compress(b"".join(rows))) + chunk(b"IEND", b"")
display(Image(data=png))
```
